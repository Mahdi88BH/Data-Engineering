### **Enriching data**

```python
new_df = pd.DataFrame(
    df['start_location_name'].value_counts().head(), 
    ).reset_index()

new_df.columns = ['address','count']

new_df
```

```python
new_df['address'] = new_df['address'].str.replace('@', 'and')

new_df['address']
```

```python
new_df[['street', 'address']] = new_df['address'].str.split(pat=',', n=1, expand=True)

new_df
```

```python
new_df = new_df[['street', 'address', 'count']]

new_df
```

```python
geo = pd.read_csv('../Data/geocodedstreet.csv')

geo
```

```python
new_df.join(other=geo, how='left', lsuffix='_new', rsuffix='_geo')
```

```python
pd.merge(new_df, geo, on='street')
```
