```python
from faker import Faker
```

```python
header = ["name", "age", "street", "city", "state","zip", "lng", "lat"]
fake = Faker()

with open('../Data/data.csv', 'w') as f:
    writer = csv.writer(f)
    writer.writerow(header)
    for r in range(1_000):
        writer.writerow([
            fake.name(),
            fake.random_int(min=18, max=80, step=1),
            fake.street_address(),
            fake.city(),
            fake.state(),
            fake.zipcode(),
            fake.longitude(),
            fake.latitude()
        ])
```
