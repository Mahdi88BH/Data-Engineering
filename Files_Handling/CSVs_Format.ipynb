## **Writing and reading CSVs**

### **Writing CSVs using the Python CSV Library**

```python
output = open('../Data/processed_data.csv', 'w')
```

```python
import csv
```

```python
my_writer = csv.writer(output, delimiter=',', quotechar='"', quoting=csv.QUOTE_MINIMAL)
```

```python
header = ["name", "age"]
my_writer.writerow(header)
```

```python
data = ["Elmahdi", "22"]
my_writer.writerow(data)
output.close()
```

```python
from faker import Faker
```

```python
header = ["name", "age", "street", "city", "state","zip", "lng", "lat"]
fake = Faker()

with open('../Data/data.csv', 'w') as f:
    writer = csv.writer(f)
    writer.writerow(header)
    for r in range(1_000):
        writer.writerow([
            fake.name(),
            fake.random_int(min=18, max=80, step=1),
            fake.street_address(),
            fake.city(),
            fake.state(),
            fake.zipcode(),
            fake.longitude(),
            fake.latitude()
        ])
```

### **Reading CSVs**

```python
with open('../Data/data.csv') as f: # By defualt in read mode
    my_reader = csv.DictReader(f) # Help to read the data by filed (not position)
    header = next(my_reader) # Get the header from the first row
    for row in my_reader:
        print(row['name'])
```
