## **Connecting to PostgreSQL with Python**

```python
import psycopg2 as db
```

```python
conn_string = "dbname='dataengineering' host='localhost' user='postgres' password='Postgres_@8899'"

conn = db.connect(conn_string)
curs = conn.cursor()
```

## **Inserting data**

```python
query = "insert into users (id, name, street, city, zip) values (1, 'John Doe', '123 Main St', 'Anytown', '12345')"
```

```python
curs.mogrify(query)
```

```python
curs.execute(query)
conn.commit()
```

## **Inserting multiple records**

```python
from faker import Faker
```

```python
fake = Faker()
data = []

for i in range(2, 1001):
    data.append((i, fake.name(), fake.street_address(), fake.city(), fake.zipcode()))
```

```python
data_db = tuple(data)
```

```python
query = "insert into users (id, name, street, city, zip) values (%s, %s, %s, %s, %s)"
```

```python
curs.executemany(query, data_db)
conn.commit()
```
