## **Definition**

**Data engineering is the development, operation, and maintenance of data infrastructure, either on-premises or in the cloud (or hybrid or multi-cloud), comprising databases and pipelines to extract, transform, and load data.**

## **Three Vs of Big Data**

The **Three Vs of Big Data** are the foundational pillars that define when a traditional database stops being enough and a true big data or data engineering architecture (like distributed storage and processing) becomes necessary. Originally coined by Gartner analyst Doug Laney, they are **Volume**, **Velocity**, and **Variety**.

Let's break down each V with a clear definition, a real-world scenario, and how it impacts data engineering.

---

## **1. Volume (Scale)**

* **What it is:** The sheer amount of data being generated and stored. We are talking about terabytes, petabytes, or even exabytes of data. Traditional relational databases (like a single MySQL or SQL Server instance on a regular machine) struggle when table sizes grow past hundreds of gigabytes or billions of rows because of storage limits and slow query scans.
* **Real-World Scenario:** Imagine an e-commerce platform like Amazon. If 10 people make a purchase, saving their transaction is easy. But what if **100 million users** click, browse, search, and buy every single day, generating gigabytes of log files every hour? Your single database hard drive will quickly run out of space, and a simple `SELECT * FROM orders` query will take hours to run.
* **Data Engineering Impact:**
    * We use **distributed storage** (like Hadoop HDFS, AWS S3, or Azure Data Lake) to split files across multiple machines.
    * We use **distributed computing frameworks** (like Apache Spark or Databricks) to process data in parallel across a cluster rather than on a single computer.

---

## **2. Velocity (Speed)**

* **What it is:** The speed at which new data is generated and the speed at which it needs to be processed, analyzed, or acted upon. Some data needs to be processed *in real-time* (streaming), while other data is processed in batches.
* **Real-World Scenario:** Think about **fraud detection for a credit card company**. If someone steals your credit card in Paris and starts making purchases, the bank cannot wait until the end of the month (or even the end of the day) to analyze the transaction. They need to analyze the data, compare it against your usual spending habits, and block the card within **milliseconds** of the swipe.
* **Data Engineering Impact:**
    * Instead of traditional **Batch Processing** (e.g., running an overnight script at 2:00 AM), data engineers build **Streaming Pipelines** using tools like Apache Kafka, Apache Flink, or Spark Streaming to ingest and react to data the exact moment it occurs.

---

## **3. Variety (Diversity of Formats)**

* **What it is:** The different forms, types, and structures that data takes. Data is no longer just neat rows and columns in an Excel sheet or SQL table.
* **Real-World Scenario:** Consider a modern healthcare app. A patient’s profile contains:
    * **Structured Data:** Age, ID number, and appointment date (stored in a SQL table).
    * **Semi-Structured Data:** JSON configuration files, clickstream logs, or API responses from wearable fitness trackers.
    * **Unstructured Data:** Doctor's voice notes (audio), X-ray scans (images), and handwritten prescription PDFs.
* **Data Engineering Impact:**
    * Traditional databases require a strict **Schema-on-Write** (you must define the table structure before saving data).
    * Big data architectures often use **Schema-on-Read** (you dump raw data into a Data Lake in its original format—JSON, CSV, Parquet, or images—and apply a schema only when you query it).

---

## **💡 Summary Table**

| V Dimension | The Core Challenge | Engineering Solution |
| --- | --- | --- |
| **Volume** | Too big for one machine/disk | Distributed storage (Data Lakes) & Processing (Spark) |
| **Velocity** | Too fast, arriving continuously | Real-time streaming pipelines (Kafka, Flink) |
| **Variety** | Too diverse (structured, semi-structured, raw) | Flexible storage formats (Parquet, Delta Lake) & Object Storage |

*(Note: Modern definitions sometimes add more Vs—like **Veracity** for data quality/trustworthiness and **Value**—but Volume, Velocity, and Variety are the core technical triad).*

## **Data engineering tools**

You have laid out a **classic, comprehensive map of the Data Engineering landscape**!

To help you master these concepts without feeling overwhelmed, let's break them down by **category, job responsibility, and where they fit into a modern production pipeline**.

---

## **1. Programming Languages: The Tooling Layer**

Language choice in Data Engineering often depends on whether you are doing **declarative query engineering**, **distributed data transformations**, or **pipeline scripting**.

* **SQL (Structured Query Language):** The single most important language in Data Engineering. Used to query, aggregate, model, and transform relational and analytical data across virtually all data stores.
* **Clojure & Groovy (JVM Languages):**
       * **Clojure:** A functional Lisp dialect running on the JVM. Excellent for immutable data processing and concurrent streaming (e.g., using libraries like *Storm* or *Cascalog*).
       * **Groovy:** A dynamic language for the JVM that simplifies Java. Historically used for script-based automation, custom Apache NiFi processors, and legacy pipeline automation.


* *(Note: In modern industry, **Python** and **Scala** are also heavy hitters alongside SQL for Apache Spark and orchestration engines).*

---

## **2. Databases: Storage & Analytics**

Databases are categorized by how they read and write data.

### **A. Operational / OLTP Databases (Row-Oriented)**

Designed for high-speed, transactional **read/write operations** (e.g., e-commerce orders, user logins).

* **Oracle & MS SQL Server:** Enterprise relational databases with strong ACID compliance and store procedures.
* **MySQL & PostgreSQL:** Open-source relational giants. *PostgreSQL* is especially popular in modern data stacks due to native JSON support and extensions like `pgvector`.

### **B. Analytical / OLAP Databases (Columnar Data Warehouses)**

Designed for massive **analytical queries** (e.g., `SUM`, `AVG`, long-term aggregations over billions of rows). They store data column-by-column rather than row-by-row.

* **Amazon Redshift:** Cloud data warehouse based on PostgreSQL, optimized for AWS ecosystem queries.
* **Google BigQuery:** Serverless, highly scalable cloud data warehouse that lets you run SQL on petabytes in seconds.

### **C. NoSQL & Distributed Data Stores**

Designed to handle **unstructured/semi-structured data**, high write throughput, or specialized search index requirements.

* **Apache Cassandra:** Wide-column NoSQL store built for extreme write volume and linear horizontal scaling across clusters without a single point of failure.
* **Elasticsearch:** A distributed search and analytics engine built on Apache Lucene, optimized for log analysis, full-text search, and time-series data.

---

## **3. Data Processing Engines: Compute & Transformation**

This is the **heavy engine room** where raw data gets cleaned, transformed, and aggregated.

```
       BATCH PROCESSING                        REAL-TIME STREAM PROCESSING
+-------------------------------+         +------------------------------------+
|  Apache Spark (RDDs / DataFrames)  |         |  Apache Flink, Apache Storm, Samza |
+-------------------------------+         +------------------------------------+

```

### **A. Batch Processing**

* **Apache Spark:** The industry standard for large-scale distributed batch and micro-batch processing.
       * **RDD (Resilient Distributed Dataset):** Spark’s fundamental low-level abstraction—an immutable, fault-tolerant collection of elements partitioned across cluster nodes.
       * *Modern Note:* Higher-level abstractions like **DataFrames** and **Spark SQL** build on top of RDDs for better optimization via Spark's Catalyst engine.



### **B. Stream Processing (Low-Latency)**

* **Apache Flink:** State-of-the-art event-driven stream processor with exact-once state management and event-time processing.
* **Apache Storm:** Early pioneer in low-latency real-time processing (processes record-by-record with minimal latency).
* **Apache Samza:** Statefull stream processing framework tightly integrated with Apache Kafka and Hadoop YARN.

---

## **4. Data Pipelines & Orchestration: Workflow Management**

Orchestrators schedule, monitor, and manage dependencies between tasks in a data pipeline.

```
+-----------------------------------------------------------------------+
|                         DAG Orchestrator                             |
|  [ Task A: Extract ] ---> [ Task B: Spark Transform ] ---> [ Task C: Load ]  |
+-----------------------------------------------------------------------+

```

* **crontab:** The simple, native Linux time-based job scheduler. Great for basic single-machine scripts, but lacks task dependency management, alerting, and retry mechanisms.
* **Apache Airflow:** The industry standard workflow orchestrator.
       * **DAG (Directed Acyclic Graph):** A collection of tasks organized to reflect their relationships and dependencies without looping back on themselves.
* **Apache NiFi:** A visual data-routing tool with an easy drag-and-drop interface, focused on **data ingestion and flow management** rather than complex analytics.
* **Luigi:** Python-based workflow engine developed by Spotify, preceding Airflow, designed to stitch together long-running batch processes.

Apache NiFi 2.x requires **Java 21 (OpenJDK 21)**.

### **Step 1: Install OpenJDK 21**

```bash
sudo apt update
sudo apt install openjdk-21-jdk -y

```

### **Step 2: Set and Verify Java Environment**

Set Java 21 as default and export `JAVA_HOME`:

```bash
sudo update-alternatives --config java
export JAVA_HOME=/usr/lib/jvm/java-21-openjdk-amd64

```

*Verification:* Run `java -version` to ensure it outputs Java 21.

---

## **Part 2: Downloading & Installing Apache NiFi**

### **Step 1: Download the Standard Binary**

Download the official Apache NiFi release archive:

```bash
wget https://dlcdn.apache.org/nifi/2.12.0/nifi-2.12.0-bin.zip

```

### **Step 2: Verify File Integrity (SHA-256)**

```bash
sha256sum nifi-2.12.0-bin.zip

```

*(Compare the hash output against the published checksum on Apache's site).*

### **Step 3: Extract & Move Files**

Extract to your preferred location (e.g., your home directory `~/` or `/opt`):

```bash
unzip nifi-2.12.0-bin.zip -d ~/
cd ~/nifi-2.12.0

```

---

## **Part 3: Essential Network Configuration (`nifi.properties`)**

Apache NiFi 2.x defaults to secure HTTPS. **Do not set both HTTP and HTTPS ports**, or NiFi will fail on startup.

1. **Open `conf/nifi.properties` in your text editor:**
```bash
nano conf/nifi.properties

```


2. **Ensure `nifi.web.http.port` is empty and `nifi.web.https.port` is set:**
```properties
nifi.web.http.port=
nifi.web.https.port=8443
nifi.web.https.host=127.0.0.1

```


3. **Save (`Ctrl+O`, `Enter`) and Exit (`Ctrl+X`).**

---

## **Part 4: Managing Credentials**

### **How to Change Login Credentials**

You can set a custom single-user username and password at any time.

1. **Stop the service (if running):**
```bash
bin/nifi.sh stop

```


2. **Set new credentials (password must be at least 12 characters):**
```bash
bin/nifi.sh set-single-user-credentials admin YourSecurePassword123!

```


3. **Start the service:**
```bash
bin/nifi.sh start

```



> **Note on Auto-Generated Credentials:** If you do not run `set-single-user-credentials`, NiFi generates a random username and password on first launch. You can retrieve them via:
```bash
grep "Generated" logs/nifi-app.log
```

## **Troubleshooting Common Errors**

* **`UnsupportedClassVersionError`**: Caused by running Java 11 instead of Java 21. Check `java -version` and reset `JAVA_HOME`.
* **`Invalid port configuration... Both nifi.web.https.port and nifi.web.http.port specified`**: Open `conf/nifi.properties` and clear `nifi.web.http.port=`.
* **Processor Warning (`Relationship success is invalid`)**: Inside the NiFi UI, every processor relationship output (`success`, `failure`) must be connected to another processor or checked as **Automatically Terminate** in processor settings.

## **Set Up Apache AirFlow**

### **Step 1: Install Prerequisites and Python Virtual Environment**

Airflow should always be installed inside an isolated Python virtual environment to avoid conflicts with your system packages.

1. **Open your terminal and install Python 3, `pip`, and the virtual environment module:**
```bash
sudo apt update
sudo apt install python3-pip python3-venv -y

```
*Verification:* Run `python3 --version` to confirm Python 3 is installed.

2. **Create a dedicated folder for Airflow and set up your virtual environment:**
```bash
python3 -m venv airflow-env
source airflow-env/bin/activate
```
*Verification:* Run `which python` to verify that it points to your virtual environment path (`/path/to/airflow-env/bin/python`).

---

### **Step 2: Install Apache Airflow**

Set your desired Airflow version and install it using the official constraints file to prevent dependency version clashes.

1. **Run the installation command:**
```bash
AIRFLOW_VERSION=2.10.2
PYTHON_VERSION="$(python3 --version | cut -d " " -f 2 | cut -d "." -f 1-2)"
CONSTRAINT_URL="https://raw.githubusercontent.com/apache/airflow/constraints-${AIRFLOW_VERSION}/constraints-${PYTHON_VERSION}.txt"

pip install "apache-airflow==${AIRFLOW_VERSION}" --constraint "${CONSTRAINT_URL}"
```

*Verification:* Run `airflow version` to confirm Airflow is installed and accessible.

---

### **Step 3: Initialize the Metadata Database**

Airflow needs a database to track the state of your tasks and DAGs. By default, it uses SQLite for local development.

1. **Set your `AIRFLOW_HOME` directory and initialize the database:**

```bash
export AIRFLOW_HOME=~/airflow
airflow db migrate

```

*Verification:* Check your home directory (`ls ~/airflow`) to confirm that `airflow.db` and configuration files have been created.

---

### **Step 4: Create an Admin User**

To log into the Airflow Web UI, you need to create an administrator account.

1. **Run the user creation command:**
```bash
airflow users create \
    --username admin \
    --firstname Admin \
    --lastname User \
    --role Admin \
    --email admin@example.com

```

*(You will be prompted to enter and confirm a secure password).*
*Verification:* Look for the success message: `Admin user admin created successfully`.

---

### **Step 5: Start Airflow Services**

Airflow runs using two primary background components: the **Webserver** (UI) and the **Scheduler** (executes tasks).

1. **Start the webserver (runs on port 8080 by default):**
```bash
airflow webserver --port 8080

```


2. **Open a **new terminal window**, activate your virtual environment, set `AIRFLOW_HOME`, and start the scheduler:**
```bash
source airflow-env/bin/activate
export AIRFLOW_HOME=~/airflow
airflow scheduler

```

*Verification:* Open your web browser and navigate to `http://localhost:8080`. Log in using your admin credentials.

---

### **Essential CLI Commands Cheat Sheet**

Here are the primary commands used to interact with Airflow from your terminal (always ensure your virtual environment is active and `AIRFLOW_HOME` is set):

| Action | Command | Description |
| --- | --- | --- |
| **Start Webserver** | `airflow webserver` | Launches the Airflow web dashboard. |
| **Start Scheduler** | `airflow scheduler` | Launches the background service that triggers tasks. |
| **List DAGs** | `airflow dags list` | Displays all available Directed Acyclic Graphs. |
| **Test a Task** | `airflow tasks test <dag_id> <task_id> <execution_date>` | Runs a single task locally for debugging purposes. |
| **Reset Database** | `airflow db reset` | Clears all metadata and resets the Airflow database. |

## **Set Up Elasticsearch**

### **Step 1: Install Prerequisites**

Elasticsearch requires Java to run. Since you installed OpenJDK 21 for NiFi earlier, your system is already prepared. You will also need `apt-transport-https` and `gnupg` to handle the official Elastic repository.

```bash
sudo apt update
sudo apt install apt-transport-https gpg curl -y
```

*Verification:* Run `java -version` to confirm Java 21 is active.

---

### **Step 2: Add Elastic GPG Key and Repository**

To download and install official packages from Elastic, add their signing key and package repository to your system:

1. **Download and import the GPG key:**
```bash
curl -fsSL https://artifacts.elastic.co/GPG-KEY-elasticsearch | sudo gpg --dearmor -o /usr/share/keyrings/elasticsearch-keyring.gpg
```


2. **Add the Elasticsearch repository to your sources list:**
```bash
echo "deb [signed-by=/usr/share/keyrings/elasticsearch-keyring.gpg] https://artifacts.elastic.co/packages/8.x/apt stable main" | sudo tee /etc/apt/sources.list.d/elasticsearch-8.x.list
```


3. **Update your package list:**
```bash
sudo apt update
```

*Verification:* Running `sudo apt update` should successfully pull package info from the new Elastic repository without errors.

---

### **Step 3: Install Elasticsearch**

Now install the package you downloaded (`deb x86_64`):

```bash
sudo apt install elasticsearch -y
```

*Verification:* Look for the output confirmation message showing that the `elasticsearch` package was successfully configured and installed.

---

### **Step 4: Configure Elasticsearch (Bind to Localhost)**

By default, Elasticsearch binds to localhost, which is ideal for local development. However, security features (like auto-generated passwords and SSL/TLS) are enabled by default in version 8.x.

1. **Open the main configuration file:**
```bash
sudo nano /etc/elasticsearch/elasticsearch.yml
```


2. **Ensure network settings point to localhost:**
```yaml
network.host: 127.0.0.1
http.port: 9200
```

3. **Save (`Ctrl+O`, `Enter`) and exit (`Ctrl+X`).**

---

### **Step 5: Start and Enable the Elasticsearch Service**

Elasticsearch runs as a systemd service on Linux.

1. **Reload systemd manager configuration and start Elasticsearch:**
```bash
sudo systemctl daemon-reload
sudo systemctl enable elasticsearch --now
```

*Verification:* Run `sudo systemctl status elasticsearch` to confirm the service status is **active (running)**.

---

### **Essential Commands for Elasticsearch**

All lifecycle commands are managed via `systemctl` and interaction is handled via `curl` requests over port `9200`.

| Action | Command | Description |
| --- | --- | --- |
| **Start Service** | `sudo systemctl start elasticsearch` | Starts the background service. |
| **Stop Service** | `sudo systemctl stop elasticsearch` | Stops the background service. |
| **Restart Service** | `sudo systemctl restart elasticsearch` | Restarts Elasticsearch. |
| **Check Status** | `sudo systemctl status elasticsearch` | Checks if the service is running. |
| **View Logs** | `sudo journalctl -u elasticsearch -f` | Streams real-time error and application logs. |

---

### **Interacting with Elasticsearch via CLI (`curl`)**

Since Elasticsearch 8 enforces security by default, you can test connectivity using an unauthenticated ping or by generating an enrollment token / password.

1. **Test basic cluster health (unauthenticated):**
```bash
curl -k -X GET "https://localhost:9200"

```

*(Note: The `-k` flag ignores self-signed SSL certificate warnings).*

2. **Generate a new bootstrap password (if prompted for authentication):**
If you need to generate credentials for users/APIs, run this command from the installation directory:
```bash
sudo /usr/share/elasticsearch/bin/elasticsearch-reset-password -u elastic
```

## **Install PostgreSQL**

### **Step 1: Install PostgreSQL**

PostgreSQL is available directly in the official Debian/Kali repositories.

1. **Update your package lists and install PostgreSQL along with its client utilities:**

```bash
sudo apt update
sudo apt install postgresql postgresql-contrib -y
```

*Verification:* Run `psql --version` to confirm it is successfully installed.

---

### **Step 2: Start and Enable the PostgreSQL Service**

PostgreSQL runs as a background systemd service on Linux.

1. **Start the service and enable it to start automatically on boot:**

```bash
sudo systemctl enable postgresql --now
```

*Verification:* Check that the service status is active:

```bash
sudo systemctl status postgresql
```

---

### **Step 3: Access the Database and Set Passwords**

By default, PostgreSQL creates a system user named `postgres` during installation. You do not log in with a regular Linux password; instead, you switch to the `postgres` user account to access the database shell.

1. **Switch to the `postgres` system user and open the interactive SQL shell (`psql`):**

```bash
sudo -u postgres psql
```

*Verification:* Your terminal prompt will change to `postgres=#`, indicating you are inside the PostgreSQL prompt.

2. **Set/Change the Password for the `postgres` Superuser:**
While inside the `psql` shell, execute this SQL command (replace `'YourSecurePassword123!'` with your desired password):

```sql
ALTER USER postgres WITH PASSWORD 'YourSecurePassword123!';
```

3. **Create a New Custom User and Database:**
It is best practice not to use the default superuser for everyday apps. Create a custom user and a dedicated database:

```sql
CREATE USER myuser WITH PASSWORD 'mypassword123';
CREATE DATABASE my_database OWNER myuser;
```


4. **Exit the `psql` prompt by typing:**

```sql
\q
```

---

### **Step 4: Create a Table in Your Database**

Now, log into your newly created database using your custom user account:

1. **Connect directly to `my_database` using `myuser`:**

```bash
psql -h localhost -U myuser -d my_database
```

*(Enter the password `mypassword123` when prompted).*

2. **Create a Sample Table (`users`):**
Run the following SQL commands inside the shell:

```sql
CREATE TABLE users (
    id SERIAL PRIMARY KEY,
    username VARCHAR(50) NOT NULL,
    email VARCHAR(100) UNIQUE NOT NULL,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);
```

*Verification:* Type `\dt` to list all tables in the database and verify that `users` is displayed.

3. **Insert Sample Data:**

```sql
INSERT INTO users (username, email) VALUES ('alice', 'alice@example.com');
```


4. **Query the Data:**

```sql
SELECT * FROM users;
```

### **Step 1: Install and Start Docker (If Not Already Installed)**

- If Docker isn't installed on your system yet, run the following commands:
```bash
sudo apt update
sudo apt install docker.io -y
```

- Make sure the Docker service is running and enabled on system startup:
```bash
sudo systemctl enable docker --now
```

*Verification:* Run `sudo systemctl status docker` to confirm the service is active and running.

---

### **Step 2: Run the pgAdmin 4 Docker Container**

- Execute the following Docker command to download and launch pgAdmin 4.

```bash
sudo docker run -d \
  --name pgadmin4 \
  -p 5050:80 \
  -e PGADMIN_DEFAULT_EMAIL=admin@admin.com \
  -e PGADMIN_DEFAULT_PASSWORD=secret \
  --restart unless-stopped \
  dpage/pgadmin4
```

#### **Breakdown of the Command flags:**

* `-d`: Runs the container in the background (detached mode).
* `--name pgadmin4`: Names the container `pgadmin4` for easy management.
* `-p 5050:80`: Maps port `5050` on your host machine to port `80` inside the container.
* `-e PGADMIN_DEFAULT_EMAIL=admin@admin.com`: Sets your initial login email.
* `-e PGADMIN_DEFAULT_PASSWORD=secret`: Sets your initial login password *(feel free to change `secret` to a secure password of your choice)*.
* `--restart unless-stopped`: Ensures the container automatically restarts whenever your computer boots up or if it crashes unexpectedly.
* `dpage/pgadmin4`: The official pre-built pgAdmin image.

*Verification:* Run `sudo docker ps` to verify that the container is up and running.

---

### **Step 3: Access pgAdmin in Your Browser**

1. **Open your web browser.**
2. **Navigate to:**
```text
http://localhost:5050
```
3. **Log in using the credentials you defined in the command:**
* **Email:** `admin@admin.com`
* **Password:** `secret` (or whatever password you set)

---

### **Step 4: Connect pgAdmin to Your Local PostgreSQL**

Because pgAdmin is running inside a Docker container, `localhost` refers to the container itself. To connect to the PostgreSQL database running on your host machine:

1. Right-click **Servers** in the left sidebar $\rightarrow$ **Create** $\rightarrow$ **Server**.
2. **General Tab:** Name your server (e.g., `Local PostgreSQL`).
3. **Connection Tab:**
* **Host name/address:** Use your host machine's Docker network gateway IP or your local IP. (Alternatively, you can use `host.docker.internal` or your machine's local IP address like `172.17.0.1` if connecting from inside Docker). 
*Tip: If you run into connection issues, use your local network IP or check your Docker bridge network IP.*
* **Port:** `5432`
* **Maintenance database:** `postgres`
* **Username:** `postgres`
* **Password:** Your PostgreSQL database password.


4. Click **Save** to connect!

### **Useful Docker Management Commands**

| Action | Command | Description |
| --- | --- | --- |
| **Stop pgAdmin** | `sudo docker stop pgadmin4` | Stops the running container. |
| **Start pgAdmin** | `sudo docker start pgadmin4` | Starts the container back up. |
| **View Logs** | `sudo docker logs pgadmin4` | Checks logs if troubleshooting is needed. |
| **Remove Container** | `sudo docker rm -f pgadmin4` | Deletes the container entirely. |
