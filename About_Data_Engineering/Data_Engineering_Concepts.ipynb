## **Definition**

**Data engineering is the development, operation, and maintenance of data infrastructure, either on-premises or in the cloud (or hybrid or multi-cloud), comprising databases and pipelines to extract, transform, and load data.**

## **Three Vs of Big Data**

The **Three Vs of Big Data** are the foundational pillars that define when a traditional database stops being enough and a true big data or data engineering architecture (like distributed storage and processing) becomes necessary. Originally coined by Gartner analyst Doug Laney, they are **Volume**, **Velocity**, and **Variety**.

Let's break down each V with a clear definition, a real-world scenario, and how it impacts data engineering.

---

## **1. Volume (Scale)**

* **What it is:** The sheer amount of data being generated and stored. We are talking about terabytes, petabytes, or even exabytes of data. Traditional relational databases (like a single MySQL or SQL Server instance on a regular machine) struggle when table sizes grow past hundreds of gigabytes or billions of rows because of storage limits and slow query scans.
* **Real-World Scenario:** Imagine an e-commerce platform like Amazon. If 10 people make a purchase, saving their transaction is easy. But what if **100 million users** click, browse, search, and buy every single day, generating gigabytes of log files every hour? Your single database hard drive will quickly run out of space, and a simple `SELECT * FROM orders` query will take hours to run.
* **Data Engineering Impact:**
    * We use **distributed storage** (like Hadoop HDFS, AWS S3, or Azure Data Lake) to split files across multiple machines.
    * We use **distributed computing frameworks** (like Apache Spark or Databricks) to process data in parallel across a cluster rather than on a single computer.

---

## **2. Velocity (Speed)**

* **What it is:** The speed at which new data is generated and the speed at which it needs to be processed, analyzed, or acted upon. Some data needs to be processed *in real-time* (streaming), while other data is processed in batches.
* **Real-World Scenario:** Think about **fraud detection for a credit card company**. If someone steals your credit card in Paris and starts making purchases, the bank cannot wait until the end of the month (or even the end of the day) to analyze the transaction. They need to analyze the data, compare it against your usual spending habits, and block the card within **milliseconds** of the swipe.
* **Data Engineering Impact:**
    * Instead of traditional **Batch Processing** (e.g., running an overnight script at 2:00 AM), data engineers build **Streaming Pipelines** using tools like Apache Kafka, Apache Flink, or Spark Streaming to ingest and react to data the exact moment it occurs.

---

## **3. Variety (Diversity of Formats)**

* **What it is:** The different forms, types, and structures that data takes. Data is no longer just neat rows and columns in an Excel sheet or SQL table.
* **Real-World Scenario:** Consider a modern healthcare app. A patient’s profile contains:
    * **Structured Data:** Age, ID number, and appointment date (stored in a SQL table).
    * **Semi-Structured Data:** JSON configuration files, clickstream logs, or API responses from wearable fitness trackers.
    * **Unstructured Data:** Doctor's voice notes (audio), X-ray scans (images), and handwritten prescription PDFs.
* **Data Engineering Impact:**
    * Traditional databases require a strict **Schema-on-Write** (you must define the table structure before saving data).
    * Big data architectures often use **Schema-on-Read** (you dump raw data into a Data Lake in its original format—JSON, CSV, Parquet, or images—and apply a schema only when you query it).

---

## **💡 Summary Table**

| V Dimension | The Core Challenge | Engineering Solution |
| --- | --- | --- |
| **Volume** | Too big for one machine/disk | Distributed storage (Data Lakes) & Processing (Spark) |
| **Velocity** | Too fast, arriving continuously | Real-time streaming pipelines (Kafka, Flink) |
| **Variety** | Too diverse (structured, semi-structured, raw) | Flexible storage formats (Parquet, Delta Lake) & Object Storage |

*(Note: Modern definitions sometimes add more Vs—like **Veracity** for data quality/trustworthiness and **Value**—but Volume, Velocity, and Variety are the core technical triad).*

## **Data engineering tools**

You have laid out a **classic, comprehensive map of the Data Engineering landscape**!

To help you master these concepts without feeling overwhelmed, let's break them down by **category, job responsibility, and where they fit into a modern production pipeline**.

---

## **1. Programming Languages: The Tooling Layer**

Language choice in Data Engineering often depends on whether you are doing **declarative query engineering**, **distributed data transformations**, or **pipeline scripting**.

* **SQL (Structured Query Language):** The single most important language in Data Engineering. Used to query, aggregate, model, and transform relational and analytical data across virtually all data stores.
* **Clojure & Groovy (JVM Languages):**
       * **Clojure:** A functional Lisp dialect running on the JVM. Excellent for immutable data processing and concurrent streaming (e.g., using libraries like *Storm* or *Cascalog*).
       * **Groovy:** A dynamic language for the JVM that simplifies Java. Historically used for script-based automation, custom Apache NiFi processors, and legacy pipeline automation.


* *(Note: In modern industry, **Python** and **Scala** are also heavy hitters alongside SQL for Apache Spark and orchestration engines).*

---

## **2. Databases: Storage & Analytics**

Databases are categorized by how they read and write data.

### **A. Operational / OLTP Databases (Row-Oriented)**

Designed for high-speed, transactional **read/write operations** (e.g., e-commerce orders, user logins).

* **Oracle & MS SQL Server:** Enterprise relational databases with strong ACID compliance and store procedures.
* **MySQL & PostgreSQL:** Open-source relational giants. *PostgreSQL* is especially popular in modern data stacks due to native JSON support and extensions like `pgvector`.

### **B. Analytical / OLAP Databases (Columnar Data Warehouses)**

Designed for massive **analytical queries** (e.g., `SUM`, `AVG`, long-term aggregations over billions of rows). They store data column-by-column rather than row-by-row.

* **Amazon Redshift:** Cloud data warehouse based on PostgreSQL, optimized for AWS ecosystem queries.
* **Google BigQuery:** Serverless, highly scalable cloud data warehouse that lets you run SQL on petabytes in seconds.

### **C. NoSQL & Distributed Data Stores**

Designed to handle **unstructured/semi-structured data**, high write throughput, or specialized search index requirements.

* **Apache Cassandra:** Wide-column NoSQL store built for extreme write volume and linear horizontal scaling across clusters without a single point of failure.
* **Elasticsearch:** A distributed search and analytics engine built on Apache Lucene, optimized for log analysis, full-text search, and time-series data.

---

## **3. Data Processing Engines: Compute & Transformation**

This is the **heavy engine room** where raw data gets cleaned, transformed, and aggregated.

```
       BATCH PROCESSING                        REAL-TIME STREAM PROCESSING
+-------------------------------+         +------------------------------------+
|  Apache Spark (RDDs / DataFrames)  |         |  Apache Flink, Apache Storm, Samza |
+-------------------------------+         +------------------------------------+

```

### **A. Batch Processing**

* **Apache Spark:** The industry standard for large-scale distributed batch and micro-batch processing.
       * **RDD (Resilient Distributed Dataset):** Spark’s fundamental low-level abstraction—an immutable, fault-tolerant collection of elements partitioned across cluster nodes.
       * *Modern Note:* Higher-level abstractions like **DataFrames** and **Spark SQL** build on top of RDDs for better optimization via Spark's Catalyst engine.



### **B. Stream Processing (Low-Latency)**

* **Apache Flink:** State-of-the-art event-driven stream processor with exact-once state management and event-time processing.
* **Apache Storm:** Early pioneer in low-latency real-time processing (processes record-by-record with minimal latency).
* **Apache Samza:** Statefull stream processing framework tightly integrated with Apache Kafka and Hadoop YARN.

---

## **4. Data Pipelines & Orchestration: Workflow Management**

Orchestrators schedule, monitor, and manage dependencies between tasks in a data pipeline.

```
+-----------------------------------------------------------------------+
|                         DAG Orchestrator                             |
|  [ Task A: Extract ] ---> [ Task B: Spark Transform ] ---> [ Task C: Load ]  |
+-----------------------------------------------------------------------+

```

* **crontab:** The simple, native Linux time-based job scheduler. Great for basic single-machine scripts, but lacks task dependency management, alerting, and retry mechanisms.
* **Apache Airflow:** The industry standard workflow orchestrator.
       * **DAG (Directed Acyclic Graph):** A collection of tasks organized to reflect their relationships and dependencies without looping back on themselves.
* **Apache NiFi:** A visual data-routing tool with an easy drag-and-drop interface, focused on **data ingestion and flow management** rather than complex analytics.
* **Luigi:** Python-based workflow engine developed by Spotify, preceding Airflow, designed to stitch together long-running batch processes.

## **Apache NIFI**

## **Part 1: Prerequisites & System Setup**

Apache NiFi 2.x requires **Java 21 (OpenJDK 21)**.

### **Step 1: Install OpenJDK 21**

```bash
sudo apt update
sudo apt install openjdk-21-jdk -y

```

### **Step 2: Set and Verify Java Environment**

Set Java 21 as default and export `JAVA_HOME`:

```bash
sudo update-alternatives --config java
export JAVA_HOME=/usr/lib/jvm/java-21-openjdk-amd64

```

*Verification:* Run `java -version` to ensure it outputs Java 21.

---

## **Part 2: Downloading & Installing Apache NiFi**

### **Step 1: Download the Standard Binary**

Download the official Apache NiFi release archive:

```bash
wget https://dlcdn.apache.org/nifi/2.12.0/nifi-2.12.0-bin.zip

```

### **Step 2: Verify File Integrity (SHA-256)**

```bash
sha256sum nifi-2.12.0-bin.zip

```

*(Compare the hash output against the published checksum on Apache's site).*

### **Step 3: Extract & Move Files**

Extract to your preferred location (e.g., your home directory `~/` or `/opt`):

```bash
unzip nifi-2.12.0-bin.zip -d ~/
cd ~/nifi-2.12.0

```

---

## **Part 3: Essential Network Configuration (`nifi.properties`)**

Apache NiFi 2.x defaults to secure HTTPS. **Do not set both HTTP and HTTPS ports**, or NiFi will fail on startup.

1. Open `conf/nifi.properties` in your text editor:
```bash
nano conf/nifi.properties

```


2. Ensure `nifi.web.http.port` is empty and `nifi.web.https.port` is set:
```properties
nifi.web.http.port=
nifi.web.https.port=8443
nifi.web.https.host=127.0.0.1

```


3. Save (`Ctrl+O`, `Enter`) and Exit (`Ctrl+X`).

---

## **Part 4: Managing Credentials**

### **How to Change Login Credentials**

You can set a custom single-user username and password at any time.

1. **Stop the service** (if running):
```bash
bin/nifi.sh stop

```


2. **Set new credentials** (password must be at least 12 characters):
```bash
bin/nifi.sh set-single-user-credentials admin YourSecurePassword123!

```


3. **Start the service**:
```bash
bin/nifi.sh start

```



> **Note on Auto-Generated Credentials:** If you do not run `set-single-user-credentials`, NiFi generates a random username and password on first launch. You can retrieve them via:
```bash
grep "Generated" logs/nifi-app.log
```

---

## **Part 5: Essential CLI Commands Cheat Sheet**

All lifecycle commands are executed via `bin/nifi.sh` from the root NiFi directory:

| Action | Command | Description |
| --- | --- | --- |
| **Start Service** | `bin/nifi.sh start` | Starts NiFi as a background service. |
| **Stop Service** | `bin/nifi.sh stop` | Gracefully shuts down the background process. |
| **Restart Service** | `bin/nifi.sh restart` | Stops and restarts the background service. |
| **Check Status** | `bin/nifi.sh status` | Checks running status and Process ID (PID). |
| **Run in Foreground** | `bin/nifi.sh run` | Launches NiFi directly in the terminal (useful for debugging startup logs). |
| **Update Credentials** | `bin/nifi.sh set-single-user-credentials <user> <pass>` | Updates authentication credentials. |

    - `https://localhost:9300/nifi`
2. Accept the self-signed SSL certificate warning in your browser.
3. Log in using your configured credentials.

## **Troubleshooting Common Errors**

* **`UnsupportedClassVersionError`**: Caused by running Java 11 instead of Java 21. Check `java -version` and reset `JAVA_HOME`.
* **`Invalid port configuration... Both nifi.web.https.port and nifi.web.http.port specified`**: Open `conf/nifi.properties` and clear `nifi.web.http.port=`.
* **Processor Warning (`Relationship success is invalid`)**: Inside the NiFi UI, every processor relationship output (`success`, `failure`) must be connected to another processor or checked as **Automatically Terminate** in processor settings.
