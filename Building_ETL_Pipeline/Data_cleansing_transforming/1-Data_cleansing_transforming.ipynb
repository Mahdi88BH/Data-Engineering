## **Step 1 – data discovery and interpretation**

```python
df_crashes = pd.read_csv('./Data/traffic_crashes.csv')

df_crashes.head()
```

## **Step 2 – checking for missing or invalid data**

```python
df_crashes.info()
```

```python
df_crashes.isnull().sum()
```

```python
df_crashes.isna().sum()
```

```python
df_crashes.dropna(axis='columns', how='all', inplace=True)
```

```python
len(df_crashes.isnull().sum())
```

```python
df_crashes = df_crashes.dropna(axis='index', thresh=2)

df_crashes.info()
```

```python
df_crashes['report_type'].value_counts()
```

```python
df_crashes['report_type'].unique()
```

```python
df_crashes = df_crashes.fillna(value={'report_type' : 'ON SCENE'})

df_crashes['report_type'].isna().sum()
```

```python
df_vehicles = pd.read_csv('./Data/traffic_crash_vehicle.csv')

df_vehicles.head()
```

```python
df = df_crashes.merge(df_vehicles, how='left', on='crash_record_id', suffixes=('_left', '_right'))

df
```

![](./Images/left-hand_merge.png)

*Visualization of a left-hand merge between two DataFrames*

```python
df_agg = df.groupby('vehicle_type').agg({'crash_record_id': 'count'}).reset_index()

df_agg
```

```python
number_of_passenger_cars_involved = df_agg[df_agg['vehicle_type'] == 'PASSENGER']['crash_record_id'].array[0]

number_of_passenger_cars_involved
```
